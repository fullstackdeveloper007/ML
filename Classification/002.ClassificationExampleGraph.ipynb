```python
import pandas as pd
df=pd.read_csv("car_ownership.csv")
df.head(3)
```

```python
from matplotlib import pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(df.monthly_salary, df.owns_car)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
colors = df['owns_car'].map({0: 'red', 1: 'green'})  # red for 0, green for 1

plt.scatter(df['monthly_salary'], df['owns_car'], c=colors, alpha=0.6, edgecolor='k')
plt.xlabel("monthly_salary")
plt.ylabel("Owns a Car (0=No, 1=Yes)")
plt.title("Car Ownership vs Salary")
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("car_ownership.csv")

plt.figure(figsize=(8, 5))
sns.stripplot(data=df, x="monthly_salary", y="owns_car", hue="owns_car", palette={0: "red", 1: "green"}, jitter=True)
plt.title("Car Ownership vs Salary")
plt.show()
```
